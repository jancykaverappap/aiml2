```python
import pandas as pd
import io

# Create a dummy scholarship.csv file if it doesn't exist
csv_data = """
Sky,AirTemp,Humidity,Wind,Water,Forecast,EnjoySport
Sunny,Warm,Normal,Strong,Warm,Same,Yes
Sunny,Warm,High,Strong,Warm,Same,Yes
Rainy,Cold,High,Strong,Warm,Change,No
Sunny,Warm,High,Strong,Cool,Change,Yes
"""

# Write the data to a file named scholarship.csv in the content directory
with open('scholarship.csv', 'w') as f:
    f.write(csv_data)

# Read dataset
data = pd.read_csv("scholarship.csv")

print("Dataset:")
print(data)

# Separate attributes and target
X = data.iloc[:, :-1]
Y = data.iloc[:, -1]

# Select only positive examples
# Ensure 'Yes' is treated as a string, as it is in the CSV data.
positive = X[Y == "Yes"]

# Initialize hypothesis with the first positive example
# Check if there are any positive examples before proceeding
if not positive.empty:
    hypothesis = list(positive.iloc[0])
else:
    print("No positive examples found. Cannot initialize hypothesis.")
    hypothesis = [] # Initialize as empty list if no positive examples

print("\nInitial Hypothesis:")
print(hypothesis)

# Find-S algorithm
# Only proceed if a hypothesis was initialized
if hypothesis:
    for i in range(1, len(positive)):
        example = list(positive.iloc[i])

        for j in range(len(hypothesis)):
            if hypothesis[j] != example[j]:
                hypothesis[j] = "?"

        print(f"\nAfter positive example {i + 1}:")
        print(hypothesis)

# Final hypothesis
print("\nFinal Hypothesis:")
print(hypothesis)
```

### Find-S Algorithm with a Different Dataset

Let's apply the Find-S algorithm to a new dataset to demonstrate its generality. This dataset will focus on whether a person will **`BuyCar`** based on `Age`, `Income`, `Student` status, and `CreditRating`.

```python
import pandas as pd
import io

# Create a new dummy CSV file for car buying prediction
car_data = """
Age,Income,Student,CreditRating,BuyCar
<=30,High,No,Fair,No
<=30,High,No,Excellent,No
31-40,High,No,Fair,Yes
>40,Medium,No,Fair,Yes
>40,Low,Yes,Fair,Yes
>40,Low,Yes,Excellent,No
31-40,Low,Yes,Excellent,Yes
<=30,Medium,No,Fair,No
<=30,Low,Yes,Fair,Yes
>40,Medium,Yes,Fair,Yes
<=30,Medium,Yes,Excellent,Yes
31-40,Medium,No,Excellent,Yes
31-40,High,Yes,Fair,Yes
>40,Medium,No,Excellent,No
"""

# Write the data to a file named car_buying.csv
with open('car_buying.csv', 'w') as f:
    f.write(car_data)

# Read the new dataset
data_car = pd.read_csv("car_buying.csv")

print("New Dataset (Car Buying):")
print(data_car)

# Separate attributes and target for the new dataset
X_car = data_car.iloc[:, :-1]
Y_car = data_car.iloc[:, -1]

# Select only positive examples for 'BuyCar' == 'Yes'
positive_car = X_car[Y_car == "Yes"]

# Initialize hypothesis with the first positive example
hypotheses_history = []
if not positive_car.empty:
    hypothesis_car = list(positive_car.iloc[0])
    hypotheses_history.append(list(hypothesis_car)) # Store h0
else:
    print("No positive examples found for car buying. Cannot initialize hypothesis.")
    hypothesis_car = []

print("\nInitial Hypothesis (Car Buying) - h0:")
print(hypothesis_car)

# Find-S algorithm for the new dataset
if hypothesis_car:
    for i in range(1, len(positive_car)):
        example_car = list(positive_car.iloc[i])

        for j in range(len(hypothesis_car)):
            if hypothesis_car[j] != example_car[j]:
                hypothesis_car[j] = "?"

        hypotheses_history.append(list(hypothesis_car)) # Store h1, h2, etc.
        print(f"\nAfter positive example {i + 1} (Car Buying) - h{i}:")
        print(hypothesis_car)

# Final hypothesis for car buying
print("\nFinal Hypothesis (Car Buying):")
print(hypothesis_car)

print("\n--- Hypothesis Evolution ---")
for idx, h in enumerate(hypotheses_history):
    print(f"h{idx}: {h}")
```

### Visualization of Hypothesis Generalization

To visualize the generalization of the hypothesis, we can plot the number of '?' attributes (which represent 'don't care' conditions) in each hypothesis (h0, h1, h2, etc.) as they evolve during the Find-S algorithm. An increasing number of '?' indicates a more general hypothesis.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the number of '?' in each hypothesis
num_wildcards = []
for h in hypotheses_history:
    count = h.count('?')
    num_wildcards.append(count)

# Create labels for the x-axis (h0, h1, h2, ...)
hypothesis_labels = [f"h{i}" for i in range(len(hypotheses_history))]

# Create the bar graph
plt.figure(figsize=(10, 6))
sns.barplot(x=hypothesis_labels, y=num_wildcards, palette="viridis")

plt.title('Evolution of Hypothesis Generalization (Number of "?" Attributes)')
plt.xlabel('Hypothesis Step')
plt.ylabel('Number of "?" Attributes')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Heatmap of Hypothesis Generalization

This heatmap visualizes the evolution of the hypothesis. A `1` (lighter color) indicates that the attribute has generalized to a '?' (don't care) value at that hypothesis step, while a `0` (darker color) indicates a specific attribute value. You can see how attributes progressively become '?' as the algorithm learns from more positive examples.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Get attribute names from the original DataFrame (X_car)
attribute_names = X_car.columns.tolist()

# Convert hypotheses_history to a DataFrame for easier manipulation
history_df = pd.DataFrame(hypotheses_history, columns=attribute_names)

# Create a numerical representation: 1 if '?', 0 otherwise
heatmap_data = history_df.applymap(lambda x: 1 if x == '?' else 0)

# Create labels for the y-axis (h0, h1, h2, ...)
hypothesis_labels_heatmap = [f"h{i}" for i in range(len(hypotheses_history))]
heatmap_data.index = hypothesis_labels_heatmap

# Create the heatmap
fig = plt.figure(figsize=(10, 7))
sns.heatmap(heatmap_data, annot=True, cmap='Blues', cbar=True, linewidths=.5)

plt.title('Heatmap of Hypothesis Generalization (0=Specific, 1=Generalized)')
plt.xlabel('Attributes')
plt.ylabel('Hypothesis Step')
plt.yticks(rotation=0) # Ensure y-axis labels are horizontal
plt.show()
```
