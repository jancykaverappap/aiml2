```python
!pip install networkx -q

import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/car/car.data"

columns = [
    "buying",
    "maint",
    "doors",
    "persons",
    "lug_boot",
    "safety",
    "class"
]

df = pd.read_csv(
    url,
    names=columns
)

print("ORIGINAL DATASET")
print(df.head())

TARGET_ATTR = "class"
TARGET_VAL = "unacc"

ATTRIBUTES = [
    "buying",
    "maint",
    "doors",
    "persons",
    "lug_boot",
    "safety"
]

DOMAIN = {
    attr: list(df[attr].dropna().unique())
    for attr in ATTRIBUTES
}

print("\nDATASET INFORMATION")
print(df.head(10))

print("\nTotal examples:", len(df))

print(
    "Positive examples:",
    (df[TARGET_ATTR] == TARGET_VAL).sum()
)

print(
    "Negative examples:",
    (df[TARGET_ATTR] != TARGET_VAL).sum()
)

print("\nCLASS DISTRIBUTION")
print(df[TARGET_ATTR].value_counts())


def covers(rule, row):

    return all(
        row[attr] == val
        for attr, val in rule.items()
    )


def evaluate(rule, examples):

    covered = examples[
        examples.apply(
            lambda r: covers(rule, r),
            axis=1
        )
    ]

    pos = (
        covered[TARGET_ATTR] == TARGET_VAL
    ).sum()

    neg = (
        covered[TARGET_ATTR] != TARGET_VAL
    ).sum()

    precision = (
        pos / (pos + neg)
        if (pos + neg) > 0
        else 0
    )

    return pos, neg, precision


def rule_to_str(rule):

    if not rule:
        return "IF (true)"

    conditions = " AND ".join(
        f"{attribute} = {value}"
        for attribute, value in rule.items()
    )

    return f"IF {conditions}"


def learn_one_rule(
        examples,
        beam_width=4,
        max_conditions=None
    ):

    if max_conditions is None:
        max_conditions = len(ATTRIBUTES)

    search_history = []

    beam = [{}]

    depth0_pos, depth0_neg, depth0_prec = evaluate(
        {},
        examples
    )

    search_history.append({
        "depth": 0,
        "candidates": [
            (
                {},
                depth0_pos,
                depth0_neg,
                depth0_prec,
                None
            )
        ],
        "beam": [{}]
    })

    best_rule = {}
    best_pos = depth0_pos
    best_neg = depth0_neg
    best_prec = depth0_prec

    for depth in range(
            1,
            max_conditions + 1
        ):

        candidates = []

        for parent in beam:

            used_attributes = set(parent.keys())

            for attr in ATTRIBUTES:

                if attr in used_attributes:
                    continue

                for value in DOMAIN[attr]:

                    child = dict(parent)

                    child[attr] = value

                    pos, neg, prec = evaluate(
                        child,
                        examples
                    )

                    if pos == 0:
                        continue

                    candidates.append(
                        (
                            child,
                            pos,
                            neg,
                            prec,
                            parent
                        )
                    )

        if not candidates:
            break

        candidates.sort(
            key=lambda c: (
                c[2] == 0,
                c[3],
                c[1]
            ),
            reverse=True
        )

        new_beam = []
        seen = set()

        for child, pos, neg, prec, parent in candidates:

            key = frozenset(
                child.items()
            )

            if key not in seen:

                seen.add(key)
                new_beam.append(child)

            if len(new_beam) >= beam_width:
                break

        beam = new_beam

        search_history.append({
            "depth": depth,
            "candidates": candidates,
            "beam": beam
        })

        pure_candidates = [
            c for c in candidates
            if c[2] == 0
        ]

        if pure_candidates:

            top = pure_candidates[0]

            best_rule = top[0]
            best_pos = top[1]
            best_neg = top[2]
            best_prec = top[3]

            break

        else:

            top = candidates[0]

            if (
                top[2] == 0,
                top[3],
                top[1]
            ) > (
                best_neg == 0,
                best_prec,
                best_pos
            ):

                best_rule = top[0]
                best_pos = top[1]
                best_neg = top[2]
                best_prec = top[3]

    return (
        best_rule,
        best_pos,
        best_neg,
        best_prec,
        search_history
    )


def sequential_covering(beam_width=4):

    working = df.copy()

    rules = []
    round_logs = []

    round_num = 1

    while (
        working[TARGET_ATTR] == TARGET_VAL
    ).sum() > 0:

        rule, pos, neg, prec, history = learn_one_rule(
            working,
            beam_width
        )

        covered_mask = working.apply(
            lambda r: covers(rule, r),
            axis=1
        )

        covered_examples = working[
            covered_mask
        ]

        covered_positive_idx = (
            covered_examples[
                covered_examples[TARGET_ATTR]
                == TARGET_VAL
            ]
            .index
            .tolist()
        )

        print("\nROUND", round_num)

        print(
            "Remaining positive examples:",
            (
                working[TARGET_ATTR]
                == TARGET_VAL
            ).sum()
        )

        print(
            "Learned Rule:",
            rule_to_str(rule),
            "THEN",
            TARGET_ATTR,
            "=",
            TARGET_VAL
        )

        print(
            "Positive covered:",
            pos
        )

        print(
            "Negative covered:",
            neg
        )

        print(
            f"Precision: {prec:.2f}"
        )

        rules.append(rule)

        round_logs.append({
            "round": round_num,
            "rule": rule,
            "pos_covered": pos,
            "neg_covered": neg,
            "covered_positive_idx": covered_positive_idx,
            "history": history
        })

        if len(covered_positive_idx) == 0:
            break

        working = working.drop(
            index=covered_positive_idx
        )

        round_num += 1

    return rules, round_logs


def plot_search_tree(
        history,
        winning_rule,
        round_num
    ):

    G = nx.DiGraph()

    labels = {}
    node_colors = []
    node_id_of_rule = {}

    def node_id(rule):

        key = frozenset(rule.items())

        if key not in node_id_of_rule:

            node_id_of_rule[key] = (
                f"n{len(node_id_of_rule)}"
            )

        return node_id_of_rule[key]

    root_pos, root_neg, root_prec = (
        history[0]["candidates"][0][1:4]
    )

    root_id = node_id({})

    G.add_node(root_id)

    labels[root_id] = (
        f"(true)\n"
        f"pos={root_pos}\n"
        f"neg={root_neg}"
    )

    winning_key = frozenset(
        winning_rule.items()
    )

    kept_keys = set()

    for level in history:

        for r in level["beam"]:

            kept_keys.add(
                frozenset(r.items())
            )

    kept_keys.add(winning_key)

    for level in history[1:]:

        for child, pos, neg, prec, parent in level["candidates"]:

            key = frozenset(child.items())

            if key not in kept_keys:
                continue

            cid = node_id(child)
            pid = node_id(parent)

            if not G.has_node(cid):

                G.add_node(cid)

                label = " AND ".join(
                    f"{a}={v}"
                    for a, v in child.items()
                )

                labels[cid] = (
                    f"{label}\n"
                    f"pos={pos} neg={neg}"
                )

            G.add_edge(pid, cid)

    for layer, nodes in enumerate(
        nx.bfs_layers(G, root_id)
    ):

        for n in nodes
```

```python
# ============================================================
# RULE-BASED CLASSIFICATION USING RIPPER, CN2 AND FOIL
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Install missing libraries
!pip install wittgenstein Orange3

# RIPPER
import wittgenstein as lw

# CN2
from Orange.data import Table, Domain, DiscreteVariable
from Orange.classification.rules import CN2Learner


# ============================================================
# 1. READ DATASET
# ============================================================

# Using the user's uploaded dataframe
df = user_df.copy()

# Renaming columns for clarity, assuming default pandas header if not provided by user
# This assumes the 'car.data' dataset structure where the last column is the target.
# Based on the previous output: ['vhigh', 'vhigh.1', '2', '2.1', 'small', 'low', 'unacc']
# Let's give them more descriptive names for the features as they are not integers.
# The column names are inferred from the first row of the 'car.data' which acts as a header
# without `header=None`. Since `header=None` was used to read, the names are generic.
# Based on common 'car.data' structure:
# buying, maint, doors, persons, lug_boot, safety, class

df.columns = ['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety', 'class_label']

print("\nDataset:")
print(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())


# ============================================================
# 2. REMOVE MISSING VALUES
# ============================================================

df = df.dropna().reset_index(drop=True)

print("\nDataset after removing missing values:")
print(df.shape)


# ============================================================
# 3. SELECT TARGET COLUMN AND DISCRETIZE
# ============================================================

# CHANGE THIS TO YOUR CLASS/TARGET COLUMN
TARGET = "class_label"

# The target variable is already categorical, so no discretization is needed.
NEW_TARGET = TARGET

X = df.drop(columns=[NEW_TARGET])
y = df[NEW_TARGET]


# ============================================================
# 4. CONVERT CATEGORICAL FEATURES TO STRINGS
# ============================================================

for col in X.columns:
    X[col] = X[col].astype(str)

y = y.astype(str)


# ============================================================
# 5. CLASS DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    x=y,
    hue=y,
    palette="Set2",
    legend=False
)

plt.title("Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Samples")
plt.tight_layout()
plt.show()


# ============================================================
# 6. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# ============================================================
# 7. RIPPER
# ============================================================

print("\n")
print("=" * 70)
print("RIPPER CLASSIFICATION")
print("=" * 70)

# Combine X and y because wittgenstein expects the target
train_ripper = X_train.copy()
train_ripper[NEW_TARGET] = y_train.values

test_ripper = X_test.copy()
test_ripper[NEW_TARGET] = y_test.values

# Create RIPPER model
ripper = lw.RIPPER(
    random_state=42,
    verbosity=0
)

# Train model
# Specify 'good' as the positive class for multi-class classification
ripper.fit(
    train_ripper,
    class_feat=NEW_TARGET,
    pos_class='good'
)

# Print learned rules
print("\nRIPPER Rules:")
print(ripper.out_model())


# ============================================================
# 8. RIPPER PREDICTION
# ============================================================

y_pred_ripper_bool = ripper.predict(X_test)

# Convert boolean predictions to string labels
# Identify the most frequent non-positive class in training data for 'False' predictions
other_classes = y_train[y_train != 'good'].value_counts()
default_class_for_false = other_classes.index[0] if not other_classes.empty else y_train.mode()[0]

y_pred_ripper = np.where(
    y_pred_ripper_bool,
    'good',
    default_class_for_false
).tolist()


print("\nRIPPER Evaluation:")
print(classification_report(
    y_test,
    y_pred_ripper,
    zero_division=0
))

ripper_accuracy = accuracy_score(y_test, y_pred_ripper)
ripper_precision = precision_score(
    y_test,
    y_pred_ripper,
    average="weighted",
    zero_division=0
)
ripper_recall = recall_score(
    y_test,
    y_pred_ripper,
    average="weighted",
    zero_division=0
)
ripper_f1 = f1_score(
    y_test,
    y_pred_ripper,
    average="weighted",
    zero_division=0
)

print("RIPPER Accuracy :", ripper_accuracy)
print("RIPPER Precision:", ripper_precision)
print("RIPPER Recall   :", ripper_recall)
print("RIPPER F1 Score :", ripper_f1)


# ============================================================
# 9. RIPPER CONFUSION MATRIX
# ============================================================

classes = sorted(y.unique())

cm_ripper = confusion_matrix(
    y_test,
    y_pred_ripper,
    labels=classes
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_ripper,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=classes,
    yticklabels=classes
)

plt.title("RIPPER Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.tight_layout()
plt.show()


# ============================================================
# 10. CN2 DATA PREPARATION
# ============================================================

print("\n")
print("=" * 70)
print("CN2 CLASSIFICATION")
print("=" * 70)

# CN2 requires discrete variables for this implementation.
# Convert all features and target to strings.

cn2_train = X_train.copy()
cn2_train[NEW_TARGET] = y_train.values

cn2_test = X_test.copy()
cn2_test[NEW_TARGET] = y_test.values

# Convert pandas data to Orange-compatible data
all_cn2 = pd.concat([cn2_train, cn2_test])

orange_data = Table.from_list(
    Domain(
        [
            DiscreteVariable.make(
                col,
                values=sorted(all_cn2[col].astype(str).unique())
            )
            for col in X.columns
        ],
        class_vars=[
            DiscreteVariable.make(
                NEW_TARGET,
                values=sorted(y.astype(str).unique())
            )
        ]
    ),
    all_cn2.astype(str).values.tolist()
)

# Separate train and test rows
train_size = len(cn2_train)

orange_train = orange_data[:train_size]
orange_test = orange_data[train_size:]


# ============================================================
# 11. TRAIN CN2
# ============================================================

cn2_learner = CN2Learner()

cn2_model = cn2_learner(orange_train)

print("\nCN2 Rules:")

for rule in cn2_model.rule_list:
    print(rule)


# ============================================================
# 12. CN2 PREDICTION
# ============================================================

predictions_cn2 = []

class_values = list(orange_train.domain.class_var.values)

for row in orange_test:

    prediction = cn2_model(row)

    try:
        predicted_class = class_values[int(prediction.value)]
    except:
        predicted_class = class_values[0]

    predictions_cn2.append(predicted_class)

y_pred_cn2 = np.array(predictions_cn2)


# ============================================================
# 13. CN2 EVALUATION
# ============================================================

print("\nCN2 Evaluation:")

print(classification_report(
    y_test,
    y_pred_cn2,
    zero_division=0
))

cn2_accuracy = accuracy_score(
    y_test,
    y_pred_cn2
)

cn2_precision = precision_score(
    y_test,
    y_pred_cn2,
    average="weighted",
    zero_division=0
)

cn2_recall = recall_score(
    y_test,
    y_pred_cn2,
    average="weighted",
    zero_division=0
)

cn2_f1 = f1_score(
    y_test,
    y_pred_cn2,
    average="weighted",
    zero_division=0
)

print("CN2 Accuracy :", cn2_accuracy)
print("CN2 Precision:", cn2_precision)
print("CN2 Recall   :", cn2_recall)
print("CN2 F1 Score :", cn2_f1)


# ============================================================
# 14. CN2 CONFUSION MATRIX
# ============================================================

cm_cn2 = confusion_matrix(
    y_test,
    y_pred_cn2,
    labels=classes
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_cn2,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=classes,
    yticklabels=classes
)

plt.title("CN2 Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.tight_layout()
plt.show()


# ============================================================
# 15. FOIL IMPLEMENTATION
# ============================================================

print("\n")
print("=" * 70)
print("FOIL FIRST-ORDER INDUCTIVE LEARNING")
print("=" * 70)


class FOIL:

    def __init__(self, target):
        self.target = target
        self.rules = []

    # --------------------------------------------------------
    # Convert a row into first-order facts
    # --------------------------------------------------------

    def create_facts(self, X, y):

        facts = []

        for index, row in X.iterrows():

            example = {
                "id": index,
                "features": row.to_dict(),
                "class": y.loc[index]
            }

            facts.append(example)

        return facts


    # --------------------------------------------------------
    # Candidate literals
    # --------------------------------------------------------

    def generate_literals(self, X):

        literals = []

        for column in X.columns:

            values = X[column].unique()

            for value in values:

                literal = (column, str(value))

                literals.append(literal)

        return literals


    # --------------------------------------------------------
    # Apply a literal to examples
    # --------------------------------------------------------

    def covers(self, example, literal):

        feature, value = literal

        return str(example["features"][feature]) == str(value)


    # --------------------------------------------------------
    # FOIL information gain
    # --------------------------------------------------------

    def foil_gain(self, positives_before,
                  negatives_before,
                  positives_after,
                  negatives_after):

        if len(positives_after) == 0:
            return -np.inf

        p0 = len(positives_before)
        n0 = len(negatives_before)

        p1 = len(positives_after)
        n1 = len(negatives_after)

        if p0 == 0 or p1 == 0:
            return -np.inf

        old_ratio = p0 / (p0 + n0)

        new_ratio = p1 / (p1 + n1)

        if old_ratio == 0 or new_ratio == 0:
            return -np.inf

        return p1 * (
            np.log2(new_ratio) -
            np.log2(old_ratio)
        )


    # --------------------------------------------------------
    # Learn rules for one class
    # --------------------------------------------------------

    def learn_class(self, examples, target_class):

        positives = [
            e for e in examples
            if e["class"] == target_class
        ]

        negatives = [
            e for e in examples
            if e["class"] != target_class
        ]

        learned_rules = []

        literals = self.generate_literals(
            pd.DataFrame([
                e["features"]
                for e in examples
            ])
        )

        while len(positives) > 0:

            current_rule = []

            current_pos = positives.copy()
            current_neg = negatives.copy()

            while len(current_neg) > 0:

                best_literal = None
                best_gain = -np.inf

                for literal in literals:

                    # Don't use the same literal twice
                    if literal in current_rule:
                        continue

                    pos_after = [
                        e for e in current_pos
                        if self.covers(e, literal)
                    ]

                    neg_after = [
                        e for e in current_neg
                        if self.covers(e, literal)
                    ]

                    gain = self.foil_gain(
                        current_pos,
                        current_neg,
                        pos_after,
                        neg_after
                    )

                    if gain > best_gain:

                        best_gain = gain
                        best_literal = literal

                if best_literal is None:
                    break

                current_rule.append(best_literal)

                current_pos = [
                    e for e in current_pos
                    if self.covers(e, best_literal)
                ]

                current_neg = [
                    e for e in current_neg
                    if self.covers(e, best_literal)
                ]

            if len(current_rule) == 0:
                break

            learned_rules.append(
                (target_class, current_rule)
            )

            # Remove positive examples covered
            # by the learned rule

            remaining_positives = []

            for example in positives:

                covered = all(
                    self.covers(example, literal)
                    for literal in current_rule
                )

                if not covered:
                    remaining_positives.append(example)

            positives = remaining_positives

        return learned_rules


    # --------------------------------------------------------
    # Learn rules for all classes
    # --------------------------------------------------------

    def fit(self, X, y):

        examples = self.create_facts(X, y)

        self.rules = []

        for target_class in sorted(y.unique()):

            rules_for_class = self.learn_class(
                examples,
                target_class
            )

            self.rules.extend(rules_for_class)

        return self.rules


    # --------------------------------------------------------
    # Predict one example
    # --------------------------------------------------------

    def predict_one(self, row, default_class):

        for target_class, rule in self.rules:

            matches = True

            for feature, value in rule:

                if str(row[feature]) != str(value):
                    matches = False
                    break

            if matches:
                return target_class

        return default_class


    # --------------------------------------------------------
    # Predict dataset
    # --------------------------------------------------------

    def predict(self, X, default_class):

        return np.array([
            self.predict_one(
                row,
                default_class
            )
            for _, row in X.iterrows()
        ])


# ============================================================
# 16. TRAIN FOIL
# ============================================================

foil = FOIL(NEW_TARGET)

foil_rules = foil.fit(
    X_train,
    y_train
)


# ============================================================
# 17. DISPLAY FOIL RULES
# ============================================================

print("\nLearned FOIL First-Order Rules:\n")

for target_class, rule in foil_rules:

    conditions = []

    for feature, value in rule:

        conditions.append(
            f"{feature}(X, {value})"
        )

    body = " AND ".join(conditions)

    print(
        f"{NEW_TARGET}(X, {target_class}) :- {body}."
    )


# ============================================================
# 18. FOIL PREDICTION
# ============================================================

default_class = y_train.mode()[0]

y_pred_foil = foil.predict(
    X_test,
    default_class
)


# ============================================================
# 19. FOIL EVALUATION
# ============================================================

print("\nFOIL Evaluation:")

print(classification_report(
    y_test,
    y_pred_foil,
    zero_division=0
))

foil_accuracy = accuracy_score(
    y_test,
    y_pred_foil
)

foil_precision = precision_score(
    y_test,
    y_pred_foil,
    average="weighted",
    zero_division=0
)

foil_recall = recall_score(
    y_test,
    y_pred_foil,
    average="weighted",
    zero_division=0
)

foil_f1 = f1_score(
    y_test,
    y_pred_foil,
    average="weighted",
    zero_division=0
)

print("FOIL Accuracy :", foil_accuracy)
print("FOIL Precision:", foil_precision)
print("FOIL Recall   :", foil_recall)
print("FOIL F1 Score :", foil_f1)


# ============================================================
# 20. FOIL CONFUSION MATRIX
# ============================================================

cm_foil = confusion_matrix(
    y_test,
    y_pred_foil,
    labels=classes
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_foil,
    annot=True,
    fmt="d",
    cmap="Oranges",
    xticklabels=classes,
    yticklabels=classes
)

plt.title("FOIL Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.tight_layout()
plt.show()


# ============================================================
# 21. MODEL PERFORMANCE COMPARISON
# ============================================================

models = [
    "RIPPER",
    "CN2",
    "FOIL"
]

accuracies = [
    ripper_accuracy,
    cn2_accuracy,
    foil_accuracy
]

precisions = [
    ripper_precision,
    cn2_precision,
    foil_precision
]

recalls = [
    ripper_recall,
    cn2_recall,
    foil_recall
]

f1_scores = [
    ripper_f1,
    cn2_f1,
    foil_f1
]


# ============================================================
# 22. ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

bars = plt.bar(
    models,
    accuracies,
    color=[
        "steelblue",
        "seagreen",
        "darkorange"
    ]
)

plt.title("Accuracy Comparison")
plt.xlabel("Rule Learning Algorithm")
plt.ylabel("Accuracy")
plt.ylim(0, 1)

for bar, value in zip(bars, accuracies):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.02,
        f"{value:.3f}",
        ha="center"
    )

plt.tight_layout()
plt.show()


# ============================================================
# 23. PRECISION / RECALL / F1 GRAPH
# ============================================================

x = np.arange(len(models))

width = 0.25

plt.figure(figsize=(10, 6))

plt.bar(
    x - width,
    precisions,
    width,
    label="Precision"
)

plt.bar(
    x,
    recalls,
    width,
    label="Recall"
)

plt.bar(
    x + width,
    f1_scores,
    width,
    label="F1 Score"
)

plt.xticks(
    x,
    models
)

plt.ylabel("Score")
plt.xlabel("Algorithm")
plt.title("Precision, Recall and F1 Comparison")

plt.ylim(0, 1)

plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# 24. CONFUSION MATRICES FOR ALL MODELS
# ============================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(18, 5)
)

sns.heatmap(
    confusion_matrix(
        y_test,
        y_pred_ripper,
        labels=classes
    ),
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=classes,
    yticklabels=classes,
    ax=axes[0]
)

axes[0].set_title("RIPPER")
axes[0].set_xlabel("Predicted")
axes[0].set_ylabel("Actual")


sns.heatmap(
    confusion_matrix(
        y_test,
        y_pred_cn2,
        labels=classes
    ),
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=classes,
    yticklabels=classes,
    ax=axes[1]
)

axes[1].set_title("CN2")
axes[1].set_xlabel("Predicted")
axes[1].set_ylabel("Actual")


sns.heatmap(
    confusion_matrix(
        y_test,
        y_pred_foil,
        labels=classes
    ),
    annot=True,
    fmt="d",
    cmap="Oranges",
    xticklabels=classes,
    yticklabels=classes,
    ax=axes[2]
)

axes[2].set_title("FOIL")
axes[2].set_xlabel("Predicted")
axes[2].set_ylabel("Actual")


plt.suptitle(
    "Confusion Matrix Comparison",
    fontsize=15
)

plt.tight_layout()
plt.show()


# ============================================================
# 25. SAVE PERFORMANCE RESULTS
# ============================================================

results = pd.DataFrame({
    "Algorithm": models,
    "Accuracy": accuracies,
    "Precision": precisions,
    "Recall": recalls,
    "F1 Score": f1_scores
})

print("\nFinal Results:")
print(results)

results.to_csv(
    "rule_learning_results.csv",
    index=False
)

print("\nResults saved to rule_learning_results.csv")


# ============================================================
# 26. SAVE FOIL RULES
# ============================================================

with open(
    "foil_rules.txt",
    "w"
) as file:

    for target_class, rule in foil_rules:

        conditions = []

        for feature, value in rule:

            conditions.append(
                f"{feature}(X, {value})"
            )

        body = " AND ".join(conditions)

        file.write(
            f"{NEW_TARGET}(X, {target_class}) :- {body}.\n"
        )

print("FOIL rules saved to foil_rules.txt")
```
