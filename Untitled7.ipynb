### Load and Preprocess Fashion MNIST Dataset

First, we'll load the Fashion MNIST dataset from TensorFlow. This dataset consists of 70,000 grayscale images of clothing items, split into 60,000 training images and 10,000 testing images. Each image is 28x28 pixels. We'll flatten the images and normalize the pixel values.

```python
import tensorflow as tf
import numpy as np
from sklearn.model_selection import train_test_split

# Load the Fashion MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

# Preprocess the data
# Normalize pixel values to be between 0 and 1
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# Flatten the images from 28x28 to 784-dimensional vectors
x_train_flat = x_train.reshape((x_train.shape[0], -1))
x_test_flat = x_test.reshape((x_test.shape[0], -1))

print(f"Training data shape: {x_train_flat.shape}")
print(f"Training labels shape: {y_train.shape}")
print(f"Test data shape: {x_test_flat.shape}")
print(f"Test labels shape: {y_test.shape}")
```

```python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import time

# Initialize a base estimator (Decision Tree)
base_estimator = DecisionTreeClassifier(random_state=42)

# Initialize Bagging Classifier
bagging_clf = BaggingClassifier(
    estimator=base_estimator,
    n_estimators=10, # Number of base estimators
    random_state=42,
    n_jobs=-1 # Use all available cores
)

print("Training Bagging Classifier...")
start_time = time.time()
bagging_clf.fit(x_train_flat, y_train)
end_time = time.time()
print(f"Bagging Classifier trained in {end_time - start_time:.2f} seconds.")

# Predict on the test set
y_pred_bagging = bagging_clf.predict(x_test_flat)

# Evaluate the Bagging Classifier
accuracy_bagging = accuracy_score(y_test, y_pred_bagging)
print(f"Bagging Classifier Accuracy: {accuracy_bagging:.4f}")
```

```python
from sklearn.ensemble import AdaBoostClassifier

# Initialize a weak base estimator (Decision Tree with max_depth=1 is a stump)
weak_base_estimator = DecisionTreeClassifier(max_depth=1, random_state=42)

# Initialize AdaBoost Classifier
adaboost_clf = AdaBoostClassifier(
    estimator=weak_base_estimator,
    n_estimators=50, # Number of boosting stages
    random_state=42,
    learning_rate=1.0
)

print("Training AdaBoost Classifier...")
start_time = time.time()
adaboost_clf.fit(x_train_flat, y_train)
end_time = time.time()
print(f"AdaBoost Classifier trained in {end_time - start_time:.2f} seconds.")

# Predict on the test set
y_pred_adaboost = adaboost_clf.predict(x_test_flat)

# Evaluate the AdaBoost Classifier
accuracy_adaboost = accuracy_score(y_test, y_pred_adaboost)
print(f"AdaBoost Classifier Accuracy: {accuracy_adaboost:.4f}")
```

```python
import pandas as pd

performance_data = {
    'Model': ['Bagging Classifier', 'AdaBoost Classifier'],
    'Accuracy': [accuracy_bagging, accuracy_adaboost]
}
performance_df = pd.DataFrame(performance_data)

display(performance_df)

if accuracy_bagging > accuracy_adaboost:
    print(f"\nBased on accuracy, the Bagging Classifier performed better than the AdaBoost Classifier on this dataset.")
elif accuracy_adaboost > accuracy_bagging:
    print(f"\nBased on accuracy, the AdaBoost Classifier performed better than the Bagging Classifier on this dataset.")
else:
    print(f"\nBoth Bagging and AdaBoost Classifiers achieved similar accuracy on this dataset.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a bar plot for model accuracy
plt.figure(figsize=(8, 6))
sns.barplot(x='Model', y='Accuracy', data=performance_df, hue='Model', palette='viridis', legend=False)
plt.title('Bagging vs. AdaBoost Classifier Accuracy')
plt.xlabel('Classifier Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1) # Accuracy is between 0 and 1

# Add accuracy values on top of the bars
for index, row in performance_df.iterrows():
    plt.text(index, row['Accuracy'] + 0.02, f'{row['Accuracy']:.4f}', color='black', ha="center")

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

# --- Bagging Classifier Metrics ---
# For multi-class classification, specify 'average' parameter. 'weighted' accounts for class imbalance.
precision_bagging = precision_score(y_test, y_pred_bagging, average='weighted')
recall_bagging = recall_score(y_test, y_pred_bagging, average='weighted')
f1_bagging = f1_score(y_test, y_pred_bagging, average='weighted')

# --- AdaBoost Classifier Metrics ---
precision_adaboost = precision_score(y_test, y_pred_adaboost, average='weighted')
recall_adaboost = recall_score(y_test, y_pred_adaboost, average='weighted')
f1_adaboost = f1_score(y_test, y_pred_adaboost, average='weighted')

# Create a DataFrame to display the metrics
metrics_data = {
    'Model': ['Bagging Classifier', 'AdaBoost Classifier'],
    'Precision': [precision_bagging, precision_adaboost],
    'Recall': [recall_bagging, recall_adaboost],
    'F1-Score': [f1_bagging, f1_adaboost]
}
metrics_df = pd.DataFrame(metrics_data)

display(metrics_df)
```
